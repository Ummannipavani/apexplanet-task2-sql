```python
import pandas as pd
import sys
import os

sys.path.append(os.path.abspath("../scripts"))

from db_utils import get_connection

engine = get_connection()
```

```python
query = """
SELECT * FROM sales
LIMIT 5;
"""

df = pd.read_sql(query, engine)

df
```

```python
query = """
SELECT SUM(`Sales`) AS Total_Sales
FROM sales;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT Branch,
SUM(`Sales`) AS Total_Sales
FROM sales
GROUP BY Branch;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT City,
SUM(`Sales`) AS Total_Sales
FROM sales
GROUP BY City;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT `Product line`,
SUM(`Sales`) AS Total_Sales
FROM sales
GROUP BY `Product line`;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT `Product line`,
AVG(Rating) AS Average_Rating
FROM sales
GROUP BY `Product line`
ORDER BY Average_Rating DESC;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT `Invoice ID`,
`Sales`
FROM sales
ORDER BY `Sales` DESC
LIMIT 10;
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT *
FROM sales
WHERE `Sales` >
(
    SELECT AVG(`Sales`)
    FROM sales
);
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT
    `Invoice ID`,
    `Sales`,
    RANK() OVER (ORDER BY `Sales` DESC) AS Sales_Rank
FROM sales;
"""

pd.read_sql(query, engine)
```
